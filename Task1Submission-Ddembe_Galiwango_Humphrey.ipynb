```python
# Question 5
# Program that stores a sentence "Python is fun" in a variable
sentence = "Python is fun"

# Program that converts this sentence to uppercase
uppercase_sentence = sentence.upper()
print("Uppercase sentence:", uppercase_sentence)

# Program that splits it into a list of words
words = uppercase_sentence.split()
print("List of words:", words)

# Program that creates a dictionary where each word is a key and its length as value
word_length_dict = {word: len(word) for word in words}
# Print the resulting dictionary
print("Word length dictionary:", word_length_dict)
```
