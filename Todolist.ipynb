```python
# Todo list program
tasks = []
def add_task(task):
    tasks.append(task)
def view_tasks():
    for idx, task in enumerate(tasks, 1):
        print(f"{idx}. {task}")
def remove_task(task_number):
    if 0 < task_number <= len(tasks):
        tasks.pop(task_number - 1)
    else:
        print("Invalid task number")  

# Example usage
# Adding tasks
print("Todo List Program")
print("------------------")
print("Adding tasks...")
add_task("Buy groceries")  
add_task("Read a book")  
# Viewing tasks
print("Current tasks:")
view_tasks()
# Removing a task
print("Removing task 1...")
remove_task(1)  
# Viewing tasks again
print("Current tasks after removal:")
view_tasks()
```
